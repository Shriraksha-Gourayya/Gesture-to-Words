```
# Force uninstall conflicting versions and reinstall a consistent set
!pip uninstall -y numpy tensorflow mediapipe opencv-python opencv-contrib-python jax jaxlib protobuf sounddevice scikit-learn
!pip install numpy==1.26.4 tensorflow==2.19.0 mediapipe==0.10.21 opencv-contrib-python==4.11.0.86 scikit-learn==1.6.1 jax==0.7.1 jaxlib==0.7.1 protobuf==4.25.8 sounddevice==0.5.3

import os
import cv2
import numpy as np
import mediapipe as mp
from tqdm import tqdm

# ============ CONFIGURATION ============
DATASET_DIR = "/content/drive/MyDrive/GestureProject/dataset"
SAVE_DIR = "/content/drive/MyDrive/GestureProject/dataset_keypoints"
SEQ_LENGTH = 30
FEATURE_DIM = 1629  # 468*3 + 33*3 + 21*3 + 21*3

mp_holistic = mp.solutions.holistic

# ============ EXTRACT KEYPOINTS ============
def extract_features(results):
    """Extract keypoints from MediaPipe results."""
    face = np.array([[lm.x, lm.y, lm.z] for lm in results.face_landmarks.landmark]).flatten() \
        if results.face_landmarks else np.zeros(468 * 3)

    pose = np.array([[lm.x, lm.y, lm.z] for lm in results.pose_landmarks.landmark]).flatten() \
        if results.pose_landmarks else np.zeros(33 * 3)

    left = np.array([[lm.x, lm.y, lm.z] for lm in results.left_hand_landmarks.landmark]).flatten() \
        if results.left_hand_landmarks else np.zeros(21 * 3)

    right = np.array([[lm.x, lm.y, lm.z] for lm in results.right_hand_landmarks.landmark]).flatten() \
        if results.right_hand_landmarks else np.zeros(21 * 3)

    return np.concatenate([face, pose, left, right])


# ============ PROCESS VIDEO ============
def process_video(video_path):
    """
    Extract keypoint sequence from video.
    Uses EVENLY SPACED frame sampling for videos of any length.
    Works with any resolution (720p, 1080p, 4K, etc.)
    """
    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        raise ValueError(f"Cannot open video: {video_path}")

    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    # Handle edge case: video has 0 frames
    if total_frames == 0:
        cap.release()
        raise ValueError(f"Video has 0 frames: {video_path}")

    # ⭐ KEY: Get evenly spaced frame indices
    if total_frames >= SEQ_LENGTH:
        # Long video: sample evenly across entire duration
        indices = np.linspace(0, total_frames - 1, SEQ_LENGTH, dtype=int)
    else:
        # Short video: use all available frames
        indices = list(range(total_frames))

    sequences = []

    with mp_holistic.Holistic(
        model_complexity=1,
        min_detection_confidence=0.5,
        min_tracking_confidence=0.5
    ) as holistic:

        frame_idx = 0
        target_idx = 0

        while cap.isOpened() and target_idx < len(indices):
            ret, frame = cap.read()
            if not ret:
                break

            # Only process frames at target indices
            if frame_idx == indices[target_idx]:
                rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                results = holistic.process(rgb)
                sequences.append(extract_features(results))
                target_idx += 1

            frame_idx += 1

    cap.release()

    # Pad with zeros if video was too short
    while len(sequences) < SEQ_LENGTH:
        sequences.append(np.zeros(FEATURE_DIM))

    return np.array(sequences[:SEQ_LENGTH])


# ============ MAIN EXTRACTION ============
os.makedirs(SAVE_DIR, exist_ok=True)

print("="*60)
print("🎬 KEYPOINT EXTRACTION")
print("="*60)
print(f"📁 Dataset: {DATASET_DIR}")
print(f"💾 Output:  {SAVE_DIR}")
print(f"📊 Sequence Length: {SEQ_LENGTH}")
print("="*60)

# Get all gestures
gestures = sorted([g for g in os.listdir(DATASET_DIR)
                   if os.path.isdir(os.path.join(DATASET_DIR, g))
                   and not g.startswith('.')])

print(f"\n📂 Found {len(gestures)} gesture classes\n")

total_processed = 0
total_skipped = 0
total_errors = 0

for gesture in gestures:
    gesture_path = os.path.join(DATASET_DIR, gesture)
    save_gesture_path = os.path.join(SAVE_DIR, gesture)
    os.makedirs(save_gesture_path, exist_ok=True)

    # Get video files
    videos = [f for f in os.listdir(gesture_path)
              if f.lower().endswith(('.mp4', '.avi', '.mov', '.mkv', '.webm'))]

    print(f"📂 {gesture} ({len(videos)} videos)")

    for video in tqdm(videos, desc=f"   Processing", leave=False):
        npy_name = os.path.splitext(video)[0] + '.npy'
        npy_path = os.path.join(save_gesture_path, npy_name)

        # Skip if already exists
        if os.path.exists(npy_path):
            total_skipped += 1
            continue

        try:
            video_path = os.path.join(gesture_path, video)
            seq = process_video(video_path)
            np.save(npy_path, seq)
            total_processed += 1
        except Exception as e:
            print(f"\n   ❌ Error: {video} - {e}")
            total_errors += 1

# ============ SUMMARY ============
print("\n" + "="*60)
print("✅ EXTRACTION COMPLETE!")
print("="*60)
print(f"   Processed: {total_processed}")
print(f"   Skipped:   {total_skipped} (already existed)")
print(f"   Errors:    {total_errors}")
print("="*60)
```

```
import os
import cv2
import numpy as np
import mediapipe as mp

DATASET_DIR = "/content/drive/MyDrive/GestureProject/dataset"
SAVE_DIR = "/content/drive/MyDrive/GestureProject/dataset_keypoints"
SEQ_LEN = 30

mp_holistic = mp.solutions.holistic

def extract_features(results):
    face = np.array([[lm.x, lm.y, lm.z] for lm in results.face_landmarks.landmark]).flatten() \
        if results.face_landmarks else np.zeros(468*3)
    pose = np.array([[lm.x, lm.y, lm.z] for lm in results.pose_landmarks.landmark]).flatten() \
        if results.pose_landmarks else np.zeros(33*3)
    left = np.array([[lm.x, lm.y, lm.z] for lm in results.left_hand_landmarks.landmark]).flatten() \
        if results.left_hand_landmarks else np.zeros(21*3)
    right = np.array([[lm.x, lm.y, lm.z] for lm in results.right_hand_landmarks.landmark]).flatten() \
        if results.right_hand_landmarks else np.zeros(21*3)

    return np.concatenate([face, pose, left, right])


def process_video(path):
    cap = cv2.VideoCapture(path)
    seq = []

    with mp_holistic.Holistic(model_complexity=1) as holistic:
        while True:
            ret, frame = cap.read()
            if not ret:
                break

            rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            results = holistic.process(rgb)
            seq.append(extract_features(results))

            if len(seq) == SEQ_LEN:
                break

    cap.release()

    while len(seq) < SEQ_LEN:
        seq.append(np.zeros(1629))

    return np.array(seq)


# -------------------------
# MAIN EXTRACTION (SKIP DONE)
# -------------------------

os.makedirs(SAVE_DIR, exist_ok=True)

for gesture in os.listdir(DATASET_DIR):
    g_path = os.path.join(DATASET_DIR, gesture)
    if not os.path.isdir(g_path):
        continue

    save_g_path = os.path.join(SAVE_DIR, gesture)
    os.makedirs(save_g_path, exist_ok=True)

    for vid in os.listdir(g_path):
        if not vid.lower().endswith((".mp4", ".avi", ".mov", ".mkv")):
            continue

        # ✨ SKIP IF ALREADY EXTRACTED
        npy_name = vid.rsplit(".", 1)[0] + ".npy"
        npy_path = os.path.join(save_g_path, npy_name)

        if os.path.exists(npy_path):
            print(f"Skipping (already exists): {gesture}/{vid}")
            continue

        print(f"Extracting {gesture}/{vid}")
        seq = process_video(os.path.join(g_path, vid))

        np.save(npy_path, seq)

print("✔ Keypoint extraction complete!")
```

```
import os

BASE = "/content/drive/MyDrive/GestureProject/dataset_keypoints"

def clean_name(name):
    name = name.strip()          # remove leading/trailing spaces
    name = name.replace(" ", "_")  # replace spaces with underscore
    return name

folders = os.listdir(BASE)

for folder in folders:
    old_path = os.path.join(BASE, folder)

    if folder == ".ipynb_checkpoints":
        # delete it completely
        import shutil
        shutil.rmtree(old_path)
        print("Deleted:", folder)
        continue

    if not os.path.isdir(old_path):
        continue

    new_name = clean_name(folder)
    new_path = os.path.join(BASE, new_name)

    if old_path != new_path:
        os.rename(old_path, new_path)
        print(f"{folder}  →  {new_name}")

print("\n✔ Folder cleaning complete! No keypoints were touched.")
```

```
import os
import numpy as np
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.optimizers import Adam

DATA_DIR = "/content/drive/MyDrive/GestureProject/dataset_keypoints"
SEQ_LENGTH = 30
FEATURE_DIM = 1629

folders = sorted([f for f in os.listdir(DATA_DIR) if os.path.isdir(os.path.join(DATA_DIR, f))])
print("Training classes:", folders)

X, y = [], []
label_map = {label: idx for idx, label in enumerate(folders)}

for gesture in folders:
    gesture_path = os.path.join(DATA_DIR, gesture)
    files = [f for f in os.listdir(gesture_path) if f.endswith(".npy")]

    for file in files:
        seq = np.load(os.path.join(gesture_path, file))
        if seq.shape == (SEQ_LENGTH, FEATURE_DIM):
            X.append(seq)
            y.append(label_map[gesture])

X = np.array(X)
y = np.array(y)

print("Dataset loaded:", X.shape, y.shape)

np.save("/content/drive/MyDrive/GestureProject/models/classes.npy", np.array(folders))

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.1, random_state=42, shuffle=True
)

model = Sequential([
    LSTM(256, return_sequences=True, input_shape=(SEQ_LENGTH, FEATURE_DIM)),
    BatchNormalization(),
    Dropout(0.3),

    LSTM(128),
    BatchNormalization(),
    Dropout(0.3),

    Dense(128, activation="relu"),
    Dense(len(folders), activation="softmax")
])

model.compile(
    optimizer=Adam(learning_rate=0.0005),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

checkpoint = ModelCheckpoint(
    "/content/drive/MyDrive/GestureProject/models/gesture_model.keras",
    monitor="val_accuracy",
    save_best_only=True,
    verbose=1
)

early_stop = EarlyStopping(
    monitor="val_accuracy",
    patience=10,
    restore_best_weights=True
)

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=32,
    callbacks=[checkpoint, early_stop]
)

print("Training finished!")
```
